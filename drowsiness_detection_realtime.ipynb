```python
import torch
import cv2
import os
import numpy as np
```

```python
path = r"yolov5\runs\train\exp7\weights\best.pt"
print(os.path.exists(path))
# I moved the file exp7 from yolov5\runs\train\exp7 to the current directory, so I can load the model without any issues.
model = torch.hub.load('ultralytics/yolov5', 'custom', path=r'exp7\weights\last.pt', force_reload=True)
```

```python
cap = cv2.VideoCapture(0) # starts webcam and stores the data in cap variable
while cap.isOpened():
    ret,fram = cap.read()
    results =model(fram)
    cv2.imshow('Drowsiness Detection', np.squeeze(results.render()))

    if cv2.waitKey(10) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```
