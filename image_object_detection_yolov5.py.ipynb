```python
import torch
from matplotlib import pyplot as plt
import numpy as np
```

```python
model = torch.hub.load('ultralytics/yolov5', 'yolov5s', pretrained=True)
```

```python
imgs = ['https://ultralytics.com/images/zidane.jpg']
results = model(imgs)
```

```python
%matplotlib inline
plt.imshow(np.squeeze(results.render()))
plt.show()
```
